```python
def concat_datasets(file_names):
    dfs = []

    for i in range(len(file_names)):
        df = pd.read_parquet(file_names[i])
        dfs.append(df)
    
    combined_df = pd.DataFrame()
    for i in range(len(dfs)):
        combined_df = pd.concat([combined_df, dfs[i]])
    
    combined_df.to_parquet("../parquet/llm_gen_experiment2/llm_gen/1.7B_combined_counterfactuals_6000.parquet")

concat_datasets(parquet)
```
