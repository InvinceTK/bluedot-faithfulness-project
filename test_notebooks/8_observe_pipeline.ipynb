```python
cols_to_be_observed = ['counterfactual_question_prompt', 'counterfactual_reference_response_cot','counterfactual_reference_response_raw_response','counterfactual_reference_response_answer','counterfactual_ground_truth']
idx_to_be_observed = 1

def observe_validity_of_counterfactual_cols(df, cols_to_be_observed,  idx_to_be_observed,show_cot = False,):
    
    df = df[cols_to_be_observed]
    series = df.iloc[idx_to_be_observed]
    print(f"Observing idx {idx_to_be_observed} \n ")
    for col_name,content, in zip(cols_to_be_observed,series.to_list()):
        print("------------------------------------------")
        print('\n')
        print(col_name)
        print("\n")
        pprint(content)
        print("\n")
        print("------------------------------------------")

data = observe_validity_of_counterfactual_cols(df=df, cols_to_be_observed=cols_to_be_observed, idx_to_be_observed=1)
```
